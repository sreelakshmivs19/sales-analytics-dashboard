```python
import pandas as pd

# Load dataset
sales = pd.read_csv('/content/Sample - Superstore.csv', encoding='latin1')

# Display first rows
print(sales.head())

# Check missing values
print(sales.isnull().sum())

# Remove duplicates
sales = sales.drop_duplicates()

# Convert date column
sales['Order Date'] = pd.to_datetime(sales['Order Date'])

# Extract month and year
sales['Month'] = sales['Order Date'].dt.month
sales['Year'] = sales['Order Date'].dt.year

# Save cleaned dataset
sales.to_csv('cleaned_sales_data.csv', index=False)

print('Data Cleaning Completed Successfully')
```

```python
total_sales = sales['Sales'].sum()
print(f"Total Sales: {total_sales}")
```

```python
total_profit = sales['Profit'].sum()
print(f"Total Profit: {total_profit}")
```

```python
top_10_products_by_sales = sales.groupby('Product Name')['Sales'].sum().nlargest(10)
print("Top 10 Products by Total Sales:")
print(top_10_products_by_sales)
```

```python
regional_sales = sales.groupby('Region')['Sales'].sum()
print("Regional Sales:")
print(regional_sales)
```

```python
monthly_sales = sales.groupby('Month')['Sales'].sum().sort_index()
print("Monthly Sales:")
print(monthly_sales)
```
